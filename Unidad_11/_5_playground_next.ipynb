```python
temas = [
    {"id_tema": 1, "nombre": "Dai Dai", "autor": "Shakira"},
    {"id_tema": 2, "nombre": "Love Yourself", "autor": "BTS"},
    {"id_tema": 3, "nombre": "Waka Waka", "autor": "Shakira"},
    {"id_tema": 4, "nombre": "Fake Love", "autor": "BTS"},
    {"id_tema": 5, "nombre": "Positions", "autor": "Ariana Grande"},
]
```

## next con map

```python
iterador = map(lambda t:t["nombre"], temas)
```

```python
next(iterador, "Fin elementos")
```

## next con filter

```python
iterador = filter(lambda t:t["autor"].lower()=="shakira", temas)
```

```python
next(iterador, "Fin elementos")
```

## next con generator

```python
iterador = (t for t in temas if t["autor"].lower()=="shakira")
```

```python
next(iterador, "Fin elementos")
```
