# 1. Diccionarios

Un **diccionario** es una estructura de datos que almacena pares **clave → valor**.

```
diccionario = {
    "clave_1": valor_1,
    "clave_2": valor_2,
}
```

Características principales:
- Los valores se acceden por **nombre** (clave), no por posición
- Las claves son **únicas** — no puede haber dos iguales
- Son **mutables** — se pueden agregar, modificar y eliminar elementos
- Desde Python 3.7+, **conservan el orden de inserción**

> 💡 También se los conoce como *arreglos asociativos*, *tablas hash* o *mapas*.


## 1.1 Creación

````python
# Forma 1: llaves literales
tema = {
    "id_tema": 1,
    "tema":    "Dai Dai",
    "autor":   "Shakira",
    "genero":  "Pop",
    "lanzamiento": 2024
}
print(tema)
````

````python
# Forma 2: función dict()
tema2 = dict(id_tema=2, tema="Swim", autor="BTS", genero="K-Pop", lanzamiento=2024)
print(tema2)
````

````python
# Forma 3: fromkeys — crear diccionario con claves fijas y valor por defecto
campos = ["id_tema", "tema", "autor", "genero", "lanzamiento"]
tema_vacio = dict.fromkeys(campos, None)
print(tema_vacio)
````

## 1.2 Acceso a elementos

````python
tema1 = {"id_tema": 1, "tema": "Dai Dai", "autor": "Shakira",
        "genero": "Pop", "lanzamiento": 2024}

# Acceso por clave directa
print(tema["autor"])          # → Shakira

# Acceso seguro con get() — retorna None si no existe (no lanza error)
print(tema.get("genero"))     # → Pop
print(tema.get("streams"))     # → None (no lanza KeyError)
print(tema.get("streams", 0))  # → 0  (valor por defecto si no existe)
````

## 1.3 Iterar diccionarios

````python
tema = {"id_tema": 1, "tema": "Dai Dai", "autor": "Shakira",
        "genero": "Pop", "lanzamiento": 2024}

# Solo claves
print("Claves:"); print(list(tema.keys()))

# Solo valores
print("Valores:"); print(list(tema.values()))

# Pares clave-valor (items)
print("Items:")
for clave, valor in tema.items():
    print(f"  {clave}: {valor}")
````

## 1.4 Modificar, agregar y eliminar

````python
tema = {"id_tema": 1, "tema": "Dai Dai", "autor": "Shakira",
        "genero": "Pop", "lanzamiento": 2024}

# Agregar un campo nuevo
tema["streams"] = 1_500_000
print("Después de agregar streams:", tema["streams"])

# Modificar un valor existente
tema["genero"] = "Latin Pop"
print("Género actualizado:", tema["genero"])

# Eliminar con del (lanza KeyError si no existe)
del tema["streams"]
print("Después de del:", "streams" in tema)

# Eliminar con pop() — retorna el valor eliminado
genero_eliminado = tema.pop("genero", "No existía")
print("Género eliminado:", genero_eliminado)
````

````python
# update() — fusionar otro diccionario
tema = {"id_tema": 1, "tema": "Dai Dai", "autor": "Shakira", "genero": "Pop"}
extra = {"genero": "Latin Pop", "streams": 2_000_000}

tema.update(extra)
print(tema)   # genero actualizado, streams agregado
````

# 2. Tuplas

Una **tupla** es una secuencia ordenada e **inmutable** de elementos.

```
tupla = (elemento_0, elemento_1, elemento_2)
```

Características:
- **Inmutable**: una vez creada, no se puede modificar
- **Ordenada**: los elementos tienen posición (índice)
- **Permite duplicados**
- Se puede acceder por índice igual que las listas
- Son más eficientes en memoria que las listas

> 💡 Usá tuplas cuando el dato no debe cambiar: coordenadas, registros históricos, retorno múltiple de funciones.


## 2.1 Creación

````python
# Tupla simple
punto = (32.73, -117.14)   # (latitud, longitud)
print(type(punto), punto)

# Tupla de un solo elemento — requiere la coma
solo_uno = (42,)
print(type(solo_uno), solo_uno)

# Sin paréntesis también es tupla (packing)
colores_rgb = 255, 128, 0
print(type(colores_rgb), colores_rgb)
````

## 2.2 Acceso e inmutabilidad

````python
rating_record = (1, 3, 5, "2024-06-01")  # (id_rating, id_usuario, id_tema, fecha)

print("id_rating:", rating_record[0])
print("id_usuario:", rating_record[1])
print("fecha:", rating_record[-1])

# Slicing (igual que listas)
print("Primeros dos:", rating_record[:2])

# Inmutabilidad — esto genera TypeError
try:
    rating_record[0] = 99
except TypeError as e:
    print(f"Error: {e}")
````

## 2.3 Unpacking — desempaquetar tuplas

````python
# Unpacking: asignar elementos de la tupla a variables con nombre
id_rating, id_usuario, id_tema, fecha = (1, 3, 5, "2024-06-01")
print(f"Rating #{id_rating}: usuario {id_usuario} calificó tema {id_tema} el {fecha}")

# Muy útil en funciones que retornan múltiples valores
def stats_ratings(ratings):
    """Retorna (mínimo, máximo, promedio) de una lista de valores."""
    valores = [r["rating"] for r in ratings]
    return min(valores), max(valores), round(sum(valores)/len(valores), 2)

ratings_ejemplo = [
    {"id_rating":1, "id_tema":1, "id_usuario":1, "rating":5, "fecha":"2024-06-01"},
    {"id_rating":2, "id_tema":1, "id_usuario":2, "rating":3, "fecha":"2024-06-02"},
    {"id_rating":3, "id_tema":1, "id_usuario":3, "rating":4, "fecha":"2024-06-03"},
]

minimo, maximo, promedio = stats_ratings(ratings_ejemplo)
print(f"min={minimo}, max={maximo}, promedio={promedio}")
````

# 3. Conjuntos (sets)

Un **conjunto** es una colección **sin duplicados** y **sin orden**.

```
conjunto = {elemento_1, elemento_2, elemento_3}
conjunto = set([lista])
```

Características:
- **Sin duplicados**: agregar un elemento ya existente no tiene efecto
- **Sin orden**: no se puede acceder por índice
- **Mutable**: se pueden agregar y quitar elementos
- Soportan operaciones de conjuntos: **unión**, **intersección**, **diferencia**

> 💡 Usá conjuntos cuando necesitás saber si algo ya existe o eliminar duplicados rápidamente.


## 3.1 Creación

````python
# Desde una lista — elimina duplicados automáticamente
generos_raw = ["Pop", "K-Pop", "Pop", "Reggaeton", "Pop", "Hip-Hop", "K-Pop"]
generos = set(generos_raw)
print("Géneros únicos:", generos)
print("Tipo:", type(generos))

# Literal
favoritos = {"Shakira", "BTS", "Bad Bunny"}
print(favoritos)

# ATENCIÓN: {} sin nada crea un diccionario vacío, no un set
vacio_dict = {}
vacio_set  = set()
print(type(vacio_dict), type(vacio_set))
````

## 3.2 Operaciones de conjunto

````python
# Usuarios que calificaron el tema 1
calificaron_t1 = {1, 2, 3, 4}

# Usuarios que calificaron el tema 2
calificaron_t2 = {3, 4, 5, 6}

# Unión — calificaron al menos uno
print("Unión:         ", calificaron_t1 | calificaron_t2)

# Intersección — calificaron ambos (colaborative filtering!)
print("Intersección:  ", calificaron_t1 & calificaron_t2)

# Diferencia — calificaron t1 pero NO t2
print("Solo en t1:    ", calificaron_t1 - calificaron_t2)

# Diferencia simétrica — calificaron uno pero no ambos
print("Diferencia sim:", calificaron_t1 ^ calificaron_t2)
````

## 3.3 Operaciones de membership y mutación

````python
usuarios_activos = {"franco", "valentín", "keyla", "pablo"}

# Verificar pertenencia (más rápido que en listas)
print("¿'keyla' está activa?", "keyla" in usuarios_activos)
print("¿'lucas' está activo?", "lucas" in usuarios_activos)

# Agregar un elemento
usuarios_activos.add("milagros")
print("Después de add:", usuarios_activos)

# Agregar varios
usuarios_activos.update(["lucas", "martin"])
print("Después de update:", usuarios_activos)

# Quitar un elemento (lanza KeyError si no existe)
usuarios_activos.discard("lucas")   # discard no lanza error si no existe
print("Después de discard:", usuarios_activos)
````

# 4. ¿Cuándo usar cada estructura?

| Pregunta | Estructura | Razón |
|---|---|---|
| ¿Necesito acceder por nombre? | **Diccionario** | clave → valor |
| ¿El dato no debe cambiar nunca? | **Tupla** | inmutable |
| ¿Necesito retornar varios valores? | **Tupla** | unpacking |
| ¿Necesito eliminar duplicados? | **Conjunto** | sin repetidos |
| ¿Necesito saber si algo ya existe (rápido)? | **Conjunto** | O(1) en membership |
| ¿Necesito orden y posición? | **Lista** | acceso por índice |

En nuestro sistema de recomendación vamos a combinar las tres:

```python
# Tema: diccionario (atributos con nombre)
tema = {"id_tema": 1, "nombre": "Dai Dai", "genero": "Pop"}

# Rating: tupla (registro histórico inmutable)
rating_tuple = (id_rating, id_usuario, id_tema, valor, fecha)

# Géneros únicos: conjunto (sin duplicados, búsqueda rápida)
generos_disponibles = {"Pop", "K-Pop", "Reggaeton", "Hip-Hop"}
```

## 📋 Resumen de métodos

### Diccionarios
| Método | Descripción |
|---|---|
| `d[key]` | Acceso por clave (KeyError si no existe) |
| `d.get(key, default)` | Acceso seguro |
| `d.keys()` / `.values()` / `.items()` | Iterar |
| `d.update(otro)` | Fusionar / actualizar |
| `d.pop(key)` | Eliminar y retornar valor |
| `d.setdefault(key, val)` | Acceder o insertar si no existe |
| `dict.fromkeys(iter, val)` | Crear con claves fijas |

### Tuplas
| Operación | Descripción |
|---|---|
| `t[i]` | Acceso por índice |
| `a, b = t` | Unpacking |
| `len(t)` | Longitud |
| `x in t` | Membership |

### Conjuntos
| Operación | Descripción |
|---|---|
| `s.add(x)` | Agregar un elemento |
| `s.update(iter)` | Agregar varios |
| `s.discard(x)` | Quitar (sin error) |
| `a \| b` | Unión |
| `a & b` | Intersección |
| `a - b` | Diferencia |
| `a ^ b` | Diferencia simétrica |
| `x in s` | Membership (O(1)) |
